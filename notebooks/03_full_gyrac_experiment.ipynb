```python
from pathlib import Path
import torch
from gyrac_pic import *
```

```python
config = make_classic_gyrac_x_smoke_config()
config.name = "full_gyrac"
config.visualization.mode = "external_and_file"
config.visualization.recording_path = Path("runs/full_gyrac/run.rrd")
```

```python
domain = CylindricalPECDomain(config.resonator.radius_m, config.resonator.length_m, config.grid.shape)
resonator = TE111CylindricalResonator(config.resonator.radius_m, config.resonator.length_m, config.resonator.frequency_hz, config.resonator.electric_field_amplitude_v_per_m, config.resonator.rf_ramp_cycles)
magnetic = RampedMirrorMagneticField(config.magnetic_field.B0_tesla, config.magnetic_field.delta_B_max_tesla, config.magnetic_field.ramp_time_s, config.magnetic_field.mirror_ratio, config.resonator.length_m)
experiment = Experiment.create(config, domain, [resonator, magnetic])
```

```python
experiment.print_configuration()
experiment.print_physical_scales()
```

```python
experiment.initialize()
```

```python
experiment.visualize_initial_state()
```

```python
experiment.run(config.num_steps)
```

```python
experiment.save_checkpoint("runs/full_gyrac/checkpoints/final.pt")
```
